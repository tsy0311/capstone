This notebook covers:

4. Bivariate analysis: features vs funding success

6. Key EDA findings

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
```

```python
df = pd.read_csv('../data/startup_cleaned.csv')
print(f'Shape: {df.shape}')
df.head()
```

## 1. Descriptive Statistics

```python
df.describe()
```

## 2. Numerical Distributions

```python
num_cols = ['funding_total_usd', 'funding_rounds', 'founded_year']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, col in zip(axes, num_cols):
    if col in df.columns:
        df[col].dropna().hist(bins=30, ax=ax, edgecolor='white')
        ax.set_title(f'Distribution of {col}')
        ax.set_xlabel(col)

plt.tight_layout()
plt.savefig('../reports/figures/num_distributions.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 3. Funding Total by Outcome

```python
if 'funding_total_usd' in df.columns:
    fig, ax = plt.subplots(figsize=(10, 6))
    df.boxplot(column='funding_total_usd', by='funding_success', ax=ax)
    ax.set_title('Funding Total by Outcome')
    ax.set_xticklabels(['Failure', 'Success'])
    plt.suptitle('')
    plt.tight_layout()
    plt.savefig('../reports/figures/funding_by_outcome.png', dpi=150, bbox_inches='tight')
    plt.show()
```

## 4. Success Rate by Country (Top 10)

```python
if 'country_code' in df.columns:
    top_countries = df['country_code'].value_counts().head(10).index
    df_top = df[df['country_code'].isin(top_countries)]
    success_rate = df_top.groupby('country_code')['funding_success'].mean().sort_values(ascending=False)

    fig, ax = plt.subplots(figsize=(12, 6))
    success_rate.plot(kind='bar', ax=ax, color='steelblue', edgecolor='white')
    ax.set_title('Funding Success Rate by Country (Top 10)')
    ax.set_ylabel('Success Rate')
    ax.tick_params(axis='x', rotation=45)
    plt.tight_layout()
    plt.savefig('../reports/figures/success_by_country.png', dpi=150, bbox_inches='tight')
    plt.show()
```

## 5. Success Rate by Market (Top 10)

```python
if 'market' in df.columns:
    top_markets = df['market'].value_counts().head(10).index
    df_mkt = df[df['market'].isin(top_markets)]
    mkt_rate = df_mkt.groupby('market')['funding_success'].mean().sort_values(ascending=False)

    fig, ax = plt.subplots(figsize=(12, 6))
    mkt_rate.plot(kind='barh', ax=ax, color='coral', edgecolor='white')
    ax.set_title('Funding Success Rate by Market (Top 10)')
    ax.set_xlabel('Success Rate')
    plt.tight_layout()
    plt.savefig('../reports/figures/success_by_market.png', dpi=150, bbox_inches='tight')
    plt.show()
```

## 6. Correlation Heatmap

```python
num_df = df.select_dtypes(include=[np.number])
if len(num_df.columns) > 1:
    plt.figure(figsize=(10, 8))
    sns.heatmap(num_df.corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
    plt.title('Correlation Matrix')
    plt.tight_layout()
    plt.savefig('../reports/figures/correlation_matrix.png', dpi=150, bbox_inches='tight')
    plt.show()
```
